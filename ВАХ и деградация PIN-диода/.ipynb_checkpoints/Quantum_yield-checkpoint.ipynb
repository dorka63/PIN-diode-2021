## 1. Зависимость ВАХ от энергии (Филипп)

### Загрузка данных

1. Названия файлов [ВАХ]:

`x v y deg z mA` — прикладываемое напряжение `x` В на энергии, соответсвующей углу монохроматора `y` градусов с током накопителя `z` мА.

2. Заголовки файлов [ВАХ]:

`Время (мс)` — абсолютное время, по нему проводится усреднение;

`Сигнал (В)` — значение напряжения с PIN-диода.

```python
def angle2energy(angle: float) -> float:
    """Перевод угла (градусы) в энергию (кэВ)"""
    
    return 12.39841 / (2 * (5.341 / np.sqrt(3)) * np.sin(np.deg2rad(angle + 0.299)))


def load_cvc_grouped(skiprows: int = 1) -> dict:
    
    """Загрузка данных по маске с извлечением параметров из имени файла"""
    
    grouped_data = defaultdict(list)
    pattern = re.compile(
        r"(\d+[\.,]?\d*)\s*v\s*(\d+[\.,]?\d*)\s*deg\s*(\d+[\.,]?\d*)\s*mA\.txt"
    )

    for file_path in glob.glob("Филипп/* v * deg * mA.txt"): # извлечение из директории Филипп
        match = pattern.search(file_path)
        if not match:
            continue

        try:
            voltage = float(match.group(1).replace(",", "."))
            energy = angle2energy(float(match.group(2).replace(",", ".")))
            beam_current = float(match.group(3).replace(",", "."))

            data = np.loadtxt(file_path, skiprows=skiprows, delimiter=",")
            mean_val = np.mean(data[:, 1])
            error_val = stats.sem(data[:, 1]) # стандартная ошибка среднего

            # нормировка: напряжение / 300 кОм 
            grouped_data[energy].append(
                [beam_current, mean_val/300, error_val/300, -voltage]
            )
        except Exception as e:
            print(f"[!] Ошибка при обработке файла {file_path}: {e}")

    return dict(sorted(grouped_data.items())) # сортировка по энергии


def build_interpolations(grouped_data: dict) -> dict:
    """Создание интерполяционных функций для каждого набора данных"""
    
    result = {}
    for energy, data_list in grouped_data.items():
        data_array = np.array(data_list)
        # сортировка по напряжению
        sort_idx = np.argsort(data_array[:, 3])
        voltages, currents, errors = (
            data_array[sort_idx, 3],
            data_array[sort_idx, 1],
            data_array[sort_idx, 2],
        )
        # интерполяционная функция; 
        # w-веса, k-степень полинома
        result[energy] = UnivariateSpline(voltages, currents, w = 1/errors, k=1)
    return result
```

```python
grouped_CVC = load_cvc_grouped()

# проверка на адекваность загрузки данных (раскомментировать для отладки)

for energy, data_list in grouped_CVC.items():
    print(f"Энергия {energy:.5f} кэВ")
    data_array = np.array(data_list)
    print(f"Ток накопителя {data_array[:, 0]} mA")
    print(f"  Фототок: {data_array[:, 1]} +/- {data_array[:, 2]} мА")
    print(f"  Напряжение: {data_array[:, 3]} В")
```

```python
interpolation_functions = build_interpolations(grouped_CVC)

fig, ax = plt.subplots() 
markers = ["o", "d", "s", "D", "^"] * 3 # разные маркеры для различимости 

for i, (energy, data_list) in enumerate(grouped_CVC.items()):
    data_array = np.array(data_list)
    interp_func = interpolation_functions[energy]
    interp_voltages = np.linspace(data_array[:, 3].min(), data_array[:, 3].max(), 100)
    interp_currents = interp_func(interp_voltages)

    color = next(ax._get_lines.prop_cycler)['color']
    
    ax.plot(interp_voltages, interp_currents, '-', color=color)
    ax.errorbar(data_array[:, 3], data_array[:, 1],
                yerr=data_array[:, 2],
                xerr=0.01 * np.abs(data_array[:, 3]) + 0.3,
                fmt=markers[i % len(markers)], ms=6,
                color=color, label=f"{energy:.4f} кэВ")

ax.set_xlabel("Напряжение, В")
ax.set_ylabel("Ток, мА")
ax.grid(alpha=0.3)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12) # легенда вне графика
plt.tight_layout()

if SAVE_FIGS:
    plt.savefig("CVC(E).png", dpi=FIG_DPI)
plt.show()
```

```python
### Перегруппировка по энергии при разных напряжениях смещения ###

def regroup_by_voltage(grouped_CVC: dict) -> dict:
    grouped_by_voltage = defaultdict(list)

    for energy, rows in grouped_CVC.items():
        for beam_current,current_norm, error, voltage in rows:
            grouped_by_voltage[voltage].append((energy, current_norm, error,beam_current))

    # сортировка по энергии
    for V in grouped_by_voltage:
        grouped_by_voltage[V] = sorted(grouped_by_voltage[V], key=lambda x: x[0])

    return dict(sorted(grouped_by_voltage.items()))

grouped_by_voltage = regroup_by_voltage(grouped_CVC)
```

```python
interpolations_V = {}

for V, arr in grouped_by_voltage.items():
    arr = np.array(arr)
    energies = arr[:, 0]
    currents = arr[:, 1]
    errors = arr[:, 2]

    interpolations_V[V] = UnivariateSpline(
        energies, currents, w=1/errors, k=1
    )

voltages_to_plot = [0, -10, -20, -30, -40, -50, -60]
fig, ax = plt.subplots()
markers = ["o", "d", "s", "D", "^"] * 3

for i, V in enumerate(voltages_to_plot):
    if V not in grouped_by_voltage:
        continue

    arr = np.array(grouped_by_voltage[V])
    energies = arr[:, 0]
    currents = arr[:, 1]
    errors = arr[:, 2]


    E_dense = np.linspace(energies.min(), energies.max(), 1000)
    I_dense = -interpolations_V[V](E_dense)

    color = next(ax._get_lines.prop_cycler)['color']

    ax.plot(E_dense, I_dense*1e3, '-', color=color)

    ax.errorbar(
        energies, -currents*1e3, yerr=errors,
        fmt=markers[i % len(markers)], ms=6,
        color=color, label=f"{V:.0f} В"
    )

ax.set_xlabel("Энергия, кэВ")
ax.set_ylabel("Ток, мкА")
ax.grid(alpha=0.3)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
from collections import OrderedDict

def load_Darwin(skiprows: int = 19) -> dict:

    """
    Загрузка файлов ширины Дарвина с сортировкой по энергии
    Возвращает OrderedDict, отсортированный по энергии
    """

    
    grouped_data = OrderedDict()
    
    pattern = re.compile(r"ш_Дарв_*(\d+[\.,]?\d*)\s*кэВ\.txt")
    energy_files = []

    # Сначала соберем все файлы и их энергии
    for file_path in glob.glob("Филипп/ш_Дарв_* кэВ.txt"):
        match = pattern.search(file_path)
        if match:
            try:
                energy_keV = float(match.group(1).replace(",", "."))
                energy_files.append((energy_keV, file_path))
            except:
                continue
    
    energy_files.sort(key=lambda x: x[0])
    
    for energy_keV, file_path in energy_files:
        try:
            data = np.loadtxt(file_path, skiprows=skiprows)

            dtheta = data[:, 0]
            denergy = data[:, 1]
            zeta = data[:, 2]
            intensity = data[:, 3]
            rocking_curve = data[:, 4]

            denergy_shifted = denergy + 1000 * energy_keV
            sort_indices = np.argsort(denergy_shifted)
            
            grouped_data[energy_keV] = {
                "dtheta": dtheta[sort_indices],
                "denergy": denergy_shifted[sort_indices],
                "zeta": zeta[sort_indices],
                "intensity": intensity[sort_indices],
                "rocking_curve": rocking_curve[sort_indices],
            }
            
        except Exception as e:
            print(f"[!] Ошибка при обработке файла {file_path}: {e}")

    return grouped_data
```

```python
D_Widths = load_Darwin()

# проверка на адекваность загрузки данных (раскомментировать для отладки)

for E, block in D_Widths.items():
    print(f"\nЭнергия {E} кэВ:")
    print("dtheta:", block["dtheta"][:5])
    print("denergy:", block["denergy"][:5])
    print("intensity:", block["intensity"][:5])
    print(len(block["intensity"]))

E = 4.53269                         
entry = D_Widths[E]                   

denergy = entry["denergy"]
intensity = entry["intensity"]
plt.figure(figsize=(8,5))
plt.plot(denergy, intensity, linewidth=1)
plt.xlabel("Energy (eV)")
plt.ylabel("Intensity (a.u.)")
plt.title(f"E0 = {E} keV")
plt.grid(True)
plt.show()
```

```python
SPEC1 = np.loadtxt("spectrum-2.txt", skiprows=2)
SPEC2 = np.loadtxt("spectrum-3.txt", skiprows=2)
flux_Be = SPEC1[:, 1] # ph/s/mr^2/0.1% BW
flux_Be_Si = SPEC2[:, 1]
flux_spec = flux_Be-flux_Be_Si
energy_spec = SPEC1[:, 0]


plt.figure(figsize=(8,4))
plt.plot(energy_spec,flux_spec*0.001*energy_spec)
plt.xlabel("Energy (eV)")
plt.ylabel("Flux(E,I) [ph/s], 1 мА")
plt.grid()
plt.show()
```

```python
# Интерполируем только FLUX!
flux_spline = UnivariateSpline(energy_spec, flux_spec, s=0)


# Словарь: voltage → (E[], beam[], signal[])
plot_data = {}

for energy_keV, rows in grouped_CVC.items():
    E_eV = energy_keV * 1000.0
    rows = np.array(rows)
    if rows.ndim == 1:
        rows = rows.reshape(1, -1)

    for r in rows:
        beam     = r[0]
        voltage  = r[3]

        # Берём flux на этой энергии
        flux_val = flux_spline(E_eV)


        # Считаем произведение
        signal = flux_val * beam

        if voltage not in plot_data:
            plot_data[voltage] = {"E": [], "beam": [], "signal": []}

        plot_data[voltage]["E"].append(E_eV)
        plot_data[voltage]["beam"].append(beam)
        plot_data[voltage]["signal"].append(signal)

# Преобразуем списки в массивы
for V in plot_data:
    idx = np.argsort(plot_data[V]["E"])
    plot_data[V]["E"]      = np.array(plot_data[V]["E"])[idx]
    plot_data[V]["beam"]   = np.array(plot_data[V]["beam"])[idx]
    plot_data[V]["signal"] = np.array(plot_data[V]["signal"])[idx]
    
plt.figure(figsize=(8,4))
for V in sorted(plot_data.keys()):
    plt.plot(plot_data[V]["E"], plot_data[V]["beam"], marker="o", label=f"{V:.0f} V")

plt.xlabel("Energy (eV)")
plt.ylabel("Beam current (mA)")
plt.grid(True)
plt.legend()
plt.tight_layout()


plt.figure(figsize=(8,4))
for V in sorted(plot_data.keys()):
    plt.plot(plot_data[V]["E"], plot_data[V]["signal"]*plot_data[V]["E"]*1e-3, marker="o", label=f"{V:.0f} V")
    print(plot_data[V]["E"])

plt.xlabel("Energy (eV)")
plt.ylabel("Flux(E,I) [ph/s]")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()

```

```python
darwin = load_Darwin()
results = []

# Для финального единого графика всех интерполированных S(E)
all_S_curves = []

for E_point, data in darwin.items():

    E_darwin = np.array(data["denergy"], float)  # eV
    curve = np.array(data["intensity"], float)

    # Интерполяция кривой Darwin
    darwin_interp = interp1d(E_darwin, curve, bounds_error=False, fill_value=0)

    # равномерная сетка для каждой кривой
    N = 1500
    E_min, E_max = E_darwin.min(), E_darwin.max()
    E_uniform = np.linspace(E_min, E_max, N)
    curve_uniform = darwin_interp(E_uniform)

    plt.figure(figsize=(8, 5))

    for V in plot_data:

        E_sig = np.array(plot_data[V]["E"], float)
        S_sig = np.array(plot_data[V]["signal"], float)

        # ph/s/0.1%bw → ph/s
        Y = S_sig * E_sig * 1e-3

        idx = np.argsort(E_sig)
        E_sorted = E_sig[idx]
        Y_sorted = Y[idx]
        
        # интерполяция потока на равномерную сетку
        
        signal_interp = interp1d(
            E_sorted, Y_sorted,
            bounds_error=False,
            fill_value=(Y_sorted[0], Y_sorted[-1]))
        
        S_uniform = signal_interp(E_uniform)

        # сохраняем для итогового общего графика
        all_S_curves.append(
            (E_uniform.copy(), S_uniform.copy(), E_point, V)
        )

        product = (curve_uniform**2) * S_uniform
        integral = simpson(product, x=E_uniform)

        results.append({
            "E_point": E_point,
            "V": V,
            "integral": integral
        })

        plt.plot(E_uniform * 1e-3, product,
                 label=f"V={V}, I={integral:.2e}")

    plt.xlabel("Energy [keV]")
    plt.ylabel("After monochromator")
    plt.title(f"E0 = {E_point:.3f} keV")
    plt.grid(True)
    plt.legend()
    plt.show()


plt.figure(figsize=(10, 6))

for E_uniform, S_uniform, E_point, V in all_S_curves:
    plt.plot(
        E_uniform, S_uniform,
        label=f"E_point={E_point:.2f} keV, V={V}"
    )

plt.xlabel("Energy [eV]")
plt.ylabel("Flux [ph/s]")
plt.grid(True)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))

for i, V in enumerate(plot_data):
    data_V = [r for r in results if r['V'] == V]

    energies = [r['E_point'] for r in data_V]
    integral = [r['integral'] for r in data_V]

    plt.scatter(energies, integral,
             marker=markers[i % len(markers)],
             linestyle='-',
             label=f"V={V} В")

plt.xlabel("Energy [eV]")
plt.ylabel("Integral [ph/s]")
plt.grid(True)
plt.tight_layout()
plt.show()
    


print("\nИТОГ")
for r in results:
    print(f"E={r['E_point']:.3f} keV,  V={r['V']},  Integral={r['integral']:.3e}")
```

```python
adjusted_results = []

for r in results:
    E_point = r['E_point']       # кэВ
    V = r['V']
    integral = r['integral']


    I_at_E = -interpolations_V[V](E_point) 
    
    integral_per_current = I_at_E*1e3 / integral
    
    adjusted_results.append({
        "E_point": E_point,
        "V": V,
        "integral": integral,
        "current": I_at_E,
        "integral_per_current": integral_per_current
    })

print("\nРЕЗУЛЬТАТЫ")
for r in adjusted_results:
    print(f"E={r['E_point']:.5f} keV, V={r['V']:>5}, Integral={r['integral']:.5e}, "
          f"Current={r['current']:.5e}, Current/Integral={r['integral_per_current']:.5e}")
    
voltages_to_plot = sorted({r['V'] for r in adjusted_results})
markers = ["o", "d", "s", "D", "^", "v", "<", ">", "p", "*"]


plt.figure(figsize=(10, 6))

for i, V in enumerate(voltages_to_plot):
    data_V = [r for r in adjusted_results if r['V'] == V]

    energies = [r['E_point'] for r in data_V]
    current = [r['current']*1e3 for r in data_V]

    plt.scatter(energies, current,
             marker=markers[i % len(markers)],
             linestyle='-',
             label=f"V={V} В")
    

plt.xlabel("Энергия, кэВ")
plt.ylabel("I_{PIN} [uС]")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))

for i, V in enumerate(voltages_to_plot):
    data_V = [r for r in adjusted_results if r['V'] == V]

    energies = [r['E_point'] for r in data_V]
    integral_per_current = [r['integral_per_current'] for r in data_V]

    plt.scatter(energies, integral_per_current,
             marker=markers[i % len(markers)],
             linestyle='-',
             label=f"V={V} В")
    

plt.xlabel("Энергия, кэВ")
plt.ylabel("I_{PIN}/Flux [uС/ph]")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
